```python
# plot multiple a few random samples from train dataset with columns ['sample_id', 'catalog_content', 'image_link', 'price']
#images are for instance: https://m.media-amazon.com/images/I/51mo8htwTHL.jpg. last part before .jpg is the image id, so the
#image can be found in ../images/{image_id}.jpg
import matplotlib.pyplot as plt

fig, axs = plt.subplots(3, 3, figsize=(10, 10))
for i in range(3):
    for j in range(3):
        idx = np.random.randint(0, len(train))
        image_id = train.iloc[idx]['image_link'].split('/')[-1].split('.')[0]
        image_path = os.path.join('../images', f'{image_id}.jpg')
        img = plt.imread(image_path)
        axs[i, j].imshow(img)
        axs[i, j].set_title(f"ID: {train.iloc[idx]['sample_id']}\nPrice: {train.iloc[idx]['price']}")
        axs[i, j].axis('off')
plt.tight_layout()
plt.show()
```
